## Encapsulation in Python

It bundles data and methods that operate on that data in a single unit(a class) while restricting direct access


### Access Modifiers in Python

- These are used to implement encapsulation by controlling the visibility and accessibility of a class's property and methods


### Level of Modifiers

1.Public
2.Protected
3.Private

1.**Public**
-default state
- these are part of class's public API and are meant to be used freely

- **Scope**:
  - Global relative to the object
  - which means accessible from anywhere(inside the class, outside the class and in subclass)

- **Syntax or convention**:
  - name (No underscore)

````python
#@ Implementation

class Person:
    def __init__(self, name):
        self.name = name  # Public Property

p = Person("Dipasha")
print(p.name) #Accessed perfectly from outside
````

1. **Protected**:
- Protected members are meant to be kept hidden from outside world but remain available to the class itself, and any classes that inherit from it(subclass)

- **Scope**:
  - The parent class and its subclasses

- **Syntax or convention**:
  - You need to use dot notation as above but you need to use single underscore ```_``` before property name.

````python
class Person:
     def __init__(self, name):
        self._name = name  # Protected property(_)

## I can either access it from inside the class or subclass.

## Can i access it from outside?
````

````python
p2 = Person("Tamang")
print(p2.name)
````

````python
p2 = Person("Tamang")
print(p2._name)  ## It works in Python, But this is not the best practise of implementing OOP.
````

1. **Private**:
- Private members are strictly hidden from outside world and subclasses.
- These are used for internal mechanics that should never be tampered with directly.

- **Scope**:
  - Strictly within the original classes

- **Syntax or convention**:
  - You cannot access them directly from outside the class, you must use public ```Getter``` and ```Setter``` metgids defined inside the class to read or to modify the private data.
  - You can make properties private using a double underscore ```__``` prefix

````python
#@ Implementation

class Person:
    def __init__(self, name, age):
        self.name = name
        self._age = age # Private Property

p1 = Person("Dipasha", 21)
print(p1.name)
print(p1.__age)
````

From the error above, yoy understand that, You cannot access Private property directly.

# Private Methods
  - You can also make methods private using the double underscore prefix.
  - This method is strictly designed for internal use, it handles background logic or validation that the outside world doesnot need to know out or interact with directly.

````python
#@ Implementation of Private Method

class Calculator:
    def __init__(self):
        self.result = 0

    def __validation(self, num):
        if num > 0:
            return True
        else:
            return False

    def add(self, num):
        if self.__validation(num):
            self.result += num
        else:
            print("Invalid Number")

calc = Calculator()
calc.add(5)
print("Iteration1:", calc.result)
````

# Name Mangling

- Name Mangling is how Python implements private properties and methods.
- When you use double underscore ```__```. Python automatically renames it internally by adding _ClassName in front

for example: ```__age``` becomes ```_Person__age``` ```__validate()``` becomes ```_Calculator__validate()```

````python
#@ Example

class Person:
    def __init__(self, name, age):
        self.name = name
        self.__age = age # Private Property

p1 = Person("Dipasha", 21)
print(p1.name)
print(p1._Person__age)  ## In this case, I don't need Getter or Setter to access private property.
````

### Operator Overloading
- It allows you to define the behaviour of operators for custom objects. You can achieve operator overloading by overriding magic method in your class.

````python
#@ Implementation
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

v1 = Vector(2, 3)
v2 = Vector(4, 5)

print(v1 + v2)
print(v1 - v2)
````
