## Loops
- For loop

- Examples

```python
## Range is a function (start, stop)
## start inclusive and stop exclusive
## range(start, stop, step)
## step means, how will you move your sequence, step = 2, the sequence will be updated by 2
## range ma, default value chai, 0 bata start hune raicha
range(5)
```

```python
## for loop
# i = 0,1,2,3,4
# 5 will not be counted as range(5) omits exlusive of 5

for i in range(5):
    print(i)
```

```python
for i in range(1, 10):
    print(i)
```

```python
for i in range(1, 10, 2):
    print(i)
```

```python
for i in range(10, 1, -2):
  print(i)
```

```python
## String
str = "Dipasha Tamang"

for i in str:
  print(i)
```

```python
## while loop
## the while loop continues to execute as long as the condition in True.

count = 0

while count<5:
    print(count)
    count = count +1
```

```python
## Loop Control Statement
## break
## break statement exits the loop prematurely

for i in range(10):
    if i==5:
        break
    print(i)
```

```python
## continue
## The continmue statement skips the current iteration and continues with the next
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
```

```python
## pass

def test():
  pass
```

```python
## The pass statement is a null operation; it does nothing.

for i in range(5):
    if i==4:
        pass
    print(i)
```

```python
## Nested Loop
## a loop inside a loop

for i in range(3):
    for j in range(2):
      # print("i:", i)
      # print("j:", j)
      print(f"i:{i} and j:{j}")
```

```python
## Calculate the sum of first N natural numbers using a while and for loop
# n = 10

n = 10
sum = 0
i = 1

while i <= n:
    sum = sum + i
    i = i + 1

print("Sum =", sum)
```

```python
## Prime number between 1 and 100 using for loop

for num in range(2, 101):
    count = 0

    for i in range(1, num + 1):
        if num % i == 0:
            count = count + 1

    if count == 2:
        print(num)
```
