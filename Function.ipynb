```python
## variable length argument in python
def add(a, b, c, d):
    return a+b+c+d

sum = add(1, 2, 3, 4)
print(sum)
```

```python
def print_numbers(*args):
    for number in args:
        print(number)
```

```python
print_numbers(1, 2, 3, 4, 5, 6, 'Elevate-X')
```

```python
def print_numbers(*dipasha):
    for number in dipasha:
        print(number)
```

```python
print_numbers("Dipasha", "Tamang")
```
