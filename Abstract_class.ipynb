## Abstract Base Class in Python

- Abstract Base Class (ABC) is used to define a commong interface that subclasses must implement.
- Goal: To provide a standarized way to test whether an object ahderes to a given specification.
- To define any abstract class, the rule is simple, i.e you must use ABC Metaclass.

i.e you need to import a class from library abc to use abstract base class.

- Characteristics:
- Cannot be instatiated: Attempt to create a n instance of abs class, then it cause an error.
- **Abstract Method**: Methods declared without an implementation (you will not write code inside). Any subclass jun chai abs class bata bancha, tesma chai you need to implement the actual code for that method.
- **Concrete Method**: Regular methods with full implementations that subclassess inherit and use directly without rewriting.

```python
#@ Example
## In python, abstract classes use the built in abc (Abstract Base Class) module.
from abc import ABC, abstractmethod
```

```python
#@ Define the Abstract class

class Animal(ABC):

    # Abstract method: Implementation empty i.e no code inside the method
    # You must implement the method in Subclasses
    @abstractmethod
    def make_sound(self):
        pass

    # Concrete method: Subclass inherit this

    def sleep(self):
        print("Sleeping...")
```

```python
#@ I will write Subclass from the ABC
class Dog(Animal):
    def make_sound(self):
        print("Bark! Bark!")


class Cat(Animal):
    def make_sound(self):
        print("Meow! Meow")
```

```python
dog = Dog()
dog.make_sound()
dog.sleep()

cat = Cat()
cat.make_sound()
cat.sleep()
```

```python
#task - 1

from abc import ABC, abstractmethod

class PaymentProcessor(ABC):

    def __init__(self, merchant_id):
        self.merchant_id = merchant_id

    @abstractmethod
    def authorize(self, amount):
        pass

    @abstractmethod
    def pay(self, amount):
        pass

    def log_transaction(self, amount, status):
        print("Merchant ID:", self.merchant_id)
        print("Amount: $", amount)
        print("Status:", status)

class EsewaProcessor(PaymentProcessor):

    def authorize(self, amount):
        if amount > 0:
            print("eSewa payment authorized")
            return True
        return False

    def pay(self, amount):
        if self.authorize(amount):
            print("eSewa payment captured")
            self.log_transaction(amount, "Success")
            return True
        return False

class KhaltiProcessor(PaymentProcessor):

    def authorize(self, amount):
        if amount > 0:
            print("Khalti payment authorized")
            return True
        return False

    def pay(self, amount):
        if self.authorize(amount):
            print("Khalti payment captured")
            self.log_transaction(amount, "Success")
            return True
        return False

```

```python
#@ Code implementation from the main.py or start.py

def checkout(processor: PaymentProcessor, total):
    processor.pay(total)
```
